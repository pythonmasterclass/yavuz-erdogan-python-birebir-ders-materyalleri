## Bölüm 1 — Fonksiyon Nedir? Neden Lazım?

**Problem:** Aynı kodu tekrar tekrar yazıyorsun.

```python
# Aynı selamlama 3 kez...
print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)

# ... birkaç satır sonra yine...
print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)
```

**Çözüm:** Bu kodu bir fonksiyona koy, isim ver, istediğin yerde çağır.

**Fonksiyonların 3 büyük faydası:**
1. **Tekrar kullanım** — Bir kez yaz, istediğin kadar çağır
2. **Düzen** — Kodun okunabilir ve organize olur
3. **Bakım** — Değişiklik yapman gerekirse tek yerde yap, her yere yansısın

````python
# Fonksiyonlar olmadan, tekrar eden kod
print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)

print() # Boşluk

print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)
````

````python
# Fonksiyonlar kullanılarak
def selamla():
    print("=" * 30)
    print("Hoş geldiniz!")
    print("=" * 30)
selamla() # Fonksiyonumuzun çağrısı
print() # Boş satır
selamla() # Fonksiyonumuzun çağrısı
print() # Boş satır
selamla() # Fonksiyonumuzun çağrısı
print() # Boş satır
selamla() # Fonksiyonumuzun çağrısı
````

## Bölüm 4 — `return`: Fonksiyondan Değer Döndür

Şimdiye kadar fonksiyonlar sadece `print()` ile ekrana yazdırdı.  
Peki ya hesapladığı sonucu **geri göndermesini** istersek?

`return` = Fonksiyonun hesapladığı değeri dışarıya verir.

**Analoji:**
- `print` → Garson yemeği pişirip sana gösterir ama vermez
- `return` → Garson yemeği pişirip tabağına koyar, sen istersin yap

```python
def fonksiyon(parametre):
    sonuc = ... hesaplama ...
    return sonuc    # Sonucu geri gönder
```

## Bölüm 5 — Varsayılan Parametreler: Opsiyonel Değerler

Bazen bir parametrenin çoğu zaman aynı değeri almasını istersin.  
**Varsayılan parametre** ile bunu sağlarsın:

```python
def fonksiyon(zorunlu, opsiyonel="varsayılan_değer"):
    ...
```

- Çağırırken `opsiyonel` verilirse → verilen değer kullanılır
- Verilmezse → varsayılan değer kullanılır

````python
def selamla(isim, mesaj="Nasılsın?"):
    print(f"Merhaba {isim}! {mesaj}")

selamla('Yavuz') # mesaj verilmedi, varsayılan kullanıldı
selamla('Yavuz', 'Hoş Geldin!') # mesaj verildi, varsayılan çiğnendi
selamla('Efekan', 'Günaydın')
selamla('Efekan')
````

````python
def siparis_olustur(urun, adet=1, indirim=0):
    fiyat = adet * 100 # basit fiyatlama
    indirimli_fiyat = fiyat * (1 - indirim / 100)
    return indirimli_fiyat

print(siparis_olustur('Laptop'))
print(siparis_olustur('Telefon', 2))
print(siparis_olustur('Kulaklık', 3, 10))
````

## Bölüm 7 — Birden Fazla Değer Döndürme

Bir fonksiyon birden fazla değer döndürebilir mi?  
Evet! Python'da `return` ile **tuple** olarak birden fazla değer döndürebilirsin.

````python
def dort_islem(a, b):
    toplam = a + b
    fark = abs(a - b)
    carpim = a * b
    return toplam, fark, carpim # 3 tane değer döndürecek ve bir tuple içinde döndürecek

sonuclar = dort_islem(12, 3)
print(sonuclar)

# index ile erişme
print('Toplam:', sonuclar[0])

# unpacking
t, f, c = dort_islem(5, 8)
print(f"Toplam: {t}, Fark: {f}, Çarpım: {c}")
````

## Bölüm 8 — Kapsam (Scope): Değişken Nerede Geçerli?

Fonksiyon içinde tanımlanan değişkenler **sadece o fonksiyonun içinde** geçerlidir.  
Buna **yerel değişken** (local variable) denir.

| Tür | Nerede Tanımlanır | Nerede Geçerli |
|-----|------------------|----------------|
| **Yerel (local)** | Fonksiyon içinde | Sadece o fonksiyon içinde |
| **Global** | Fonksiyon dışında | Her yerde |

````python
def fonksiyon():
    x = 10
    print('fonksiyonun içinde', x)

fonksiyon()
print('fonksiyonun dışında', x)
````

````python
mesaj = "Ben globalim"
def fonksiyon2():
    print('fonksiyonun içinde', mesaj)
fonksiyon2()
print('fonksiyonun dışında', mesaj)
````

````python
# Aynı isimde yerel ve global değişken
sayi= 100
def degistir():
    sayi = 999
    print(f"Fonksiyonun içinde 'sayi' = {sayi}")
degistir()
print(f"Fonksiyonun dışında 'sayi' = {sayi}")
````

## Bölüm 9 — Fonksiyon İçinde Fonksiyon Çağırma

Bir fonksiyon başka bir fonksiyonu çağırabilir.  
Bu sayede karmaşık işleri küçük parçalara bölebilirsin.

````python
# Küçük, tek işli fonksiyonlar
def kare_al(sayi1):
    return sayi1 ** 2

def kup_al(sayi2):
    return sayi2 ** 3

def mutlak_deger(sayi3):
    if sayi3 < 0:
        return -sayi3
    else:
        return f"{sayi3} -> (Verdiğiniz sayı sıfır veya pozitif olduğundan dolayı herhangi bir işlem yapılmamıştır.)"

def sayi_raporu(sayi4):
    print(f"Sayı: {sayi4}")
    print(f"Karesi: {kare_al(sayi4)}")
    print(f"Küpü: {kup_al(sayi4)}")
    print(f"Mutlak Değeri: {mutlak_deger(sayi4)}")

sayi_raporu(-5)
sayi_raporu(16)
````

````python
def harf_ver(puan):
    if puan >= 90: return "AA"
    elif puan >= 80: return "BA"
    elif puan >= 70: return "BB"
    elif puan >= 60: return "CB"
    elif puan >= 50: return "CC"
    else: return "FF"
def gecti_mi(puan):
    return puan >= 50

def ogrenci_durumu(isim, vize, final):
    ortalama = vize * 0.4 + final * 0.6
    harf = harf_ver(ortalama)
    durum = "Geçti" if gecti_mi(ortalama) else "Kaldı"
    # [Doğruysa olacak iş, değer] if [koşul] else [Yanlışsa olacak iş, değer]
    # ÖDEV: Python Masterclass Ternary Yapısı videosu izlenecek.
    print(f"Öğrenci: {isim}")
    print(f"Vize: {vize}")
    print(f"Final: {final}")
    print(f"Ortalama: {ortalama:.1f}")
    print(f"Harf Notu: {harf}")
    print(f"Durum: {durum}")
    print()

ogrenci_durumu("Yavuz", 75, 85)
ogrenci_durumu("Ali", 40, 45)
ogrenci_durumu("Zeynep", 90, 95)
````

## Bölüm 11 — Fonksiyon ile Liste İşlemleri

Fonksiyonlara liste gönderip, liste döndürebilirsin.  
Bu sayede kendi `max()`, `sum()`, `sorted()` fonksiyonlarını yazabilirsin!

````python
def ortalama_hesapla(liste):
    toplam = 0
    for sayi in liste:
        toplam += sayi
    return toplam / len(liste)

notlar = [15, 75, 34, 67, 91, 43, 84]
ort = ortalama_hesapla(notlar)
print(ort)
````

````python
def pozitif_filtrele(liste):
    sonuc = []
    for sayi in liste:
        if sayi > 0:
            sonuc.append(sayi)
    return sonuc

def ciftleri_filtrele(liste):
    sonuc = []
    for sayi in liste:
        if sayi % 2 == 0:
            sonuc.append(sayi)
    return sonuc

sayilar = [-5, 3, -1, 8, 0 , -2, 7, 4, -9, 6]
print(f"Orijinal: {sayilar}")
print(f"Pozitifler: {pozitif_filtrele(sayilar)}")
print(f"Çiftler: {ciftleri_filtrele(sayilar)}")
````
