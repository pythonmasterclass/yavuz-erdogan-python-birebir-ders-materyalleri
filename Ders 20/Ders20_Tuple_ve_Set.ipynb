## Bölüm 1 — Tuple Nedir? Liste ile Farkı Ne?

**Tuple** = Sıralı, değiştirilemez (immutable) veri yapısı.

| Özellik | Liste | Tuple |
|---------|-------|-------|
| Yazım | `[1, 2, 3]` | `(1, 2, 3)` |
| Değiştirilebilir mi? | Evet (mutable) | Hayır (immutable) |
| Eleman ekleme/silme | `append`, `remove`, `pop` | Yapılamaz |
| Hız | Daha yavaş | Daha hızlı |
| Dict anahtarı olabilir mi? | Hayır | Evet |

```python
# Liste — köşeli parantez
renkler_liste = ["kırmızı", "mavi", "yeşil"]

# Tuple — normal parantez
renkler_tuple = ("kırmızı", "mavi", "yeşil")
```

Tuple'ın asıl gücü: **"Bu veri değişmemeli"** dediğin yerlerde kullanırsın.

````python
# Liste vs Tuple
renkler_liste = ["kırmızı", "mavi", "yeşil"]
renkler_tuple = ("kırmızı", "mavi", "yeşil")
print(f"Liste: {renkler_liste}")
print(f"Veri Tipi: {type(renkler_liste)}")
print(f"Tuple: {renkler_tuple}")
print(f"Veri Tipi: {type(renkler_tuple)}")
print()
# Listeyi değiştirebilirsin
renkler_liste[0] = "sarı"
print(f"Liste Değişti: {renkler_liste}")
# Tuple'ı değiştirmeye çalışalım - HATA ALIRIZ
try:
    renkler_tuple[0] = "sarı"
except TypeError as e: # hatayı gösteriyor
    print(f"Tuple Değiştirilemez {e}")
````

## Bölüm 3 — Tuple Unpacking: Tek Satırda Birden Fazla Değişken

Tuple'ın en güçlü özelliklerinden biri **unpacking** (çözme):  
Bir tuple'ın elemanlarını ayrı ayrı değişkenlere atayabilirsin.

```python
# Klasik yol
kisi = ("Yavuz", 20, "İstanbul")
isim = kisi[0]
yas = kisi[1]
sehir = kisi[2]

# Unpacking ile — tek satır!
isim, yas, sehir = ("Yavuz", 20, "İstanbul")
```

Eleman sayısı eşleşmeli! Aksi halde hata alırsın.

````python
koordinat = (38.68, 29.41)
enlem, boylam = koordinat
print(f"Enlem: {enlem}, Boylam: {boylam}")
````

````python
kisi = ("Yavuz", 20, "Bilgisayar Mühendisliği")
isim, yas, bolum = kisi
print(f"{isim}, {yas} yaşında, {bolum} öğrencisi.")
````

````python
a = 10
b = 20
print(f"Önce: a={a}, b={b}")
a, b = b, a
print(f"Sonra: a={a}, b={b}")
````

````python
notlar = (95, 88, 31, 69, 72, 45, 80)
en_yuksek, *ortalar, en_dusuk = notlar
print(f"En Yüksek: {en_yuksek}")
print(f"En Düşük: {en_dusuk}")
print(f"Kalanlar: {ortalar}")
# NOT: Tuple sıralı değil, burada sadece pozisyona göre çalışır
# Doğru kullanım için sıralanmış veri lazım
print()
notlar_sirali = tuple(sorted(notlar, reverse=True))
print(f"Sıralı Notlar: {notlar_sirali}")

birinci, ikinci, *kalanlar = notlar_sirali
print(f"En Yüksek: {birinci}")
print(f"En Yüksek İkinci: {ikinci}")
print(f"Kalanlar: {kalanlar}")
````

## Bölüm 4 — Tuple Metotları ve İşlemleri

Tuple değiştirilemez olduğu için sadece **2 metodu** var:

| Metot | Açıklama | Örnek |
|-------|----------|-------|
| `count(x)` | x kaç kez geçiyor | `(1,2,1).count(1)` → 2 |
| `index(x)` | x'in ilk indexi | `(1,2,3).index(2)` → 1 |

Ama tuple üzerinde kullanabileceğin **built-in fonksiyonlar** çok:

````python
notlar = (85, 90, 72, 90, 65, 88, 90)

# İki metot
print(f"'90' kaç kez var? {notlar.count(90)}")
print(f"'90' kaçındı indexte? {notlar.index(90)}")
print()

# Built-in fonksiyonlar
print(f"Eleman sayısı {len(notlar)}")
print(f"En büyük sayı {max(notlar)}")
print(f"En küçük sayı {min(notlar)}")
print(f"Sayıların toplamı {sum(notlar)}")
print(f"Sayıların ortalaması {sum(notlar) / len(notlar):.1f}")

# sorted() fonksiyonu -> Tuple'ı sıralar ama LİSTE döndürür.
sirali = sorted(notlar)
print(f"\nSıralı: {sirali}, Veri Tipi: {type(sirali)}")
# Tuple olarak istersek
sirali_tuple = tuple(sirali)
print(f"\nSıralı Tuple: {sirali_tuple}, Veri Tipi: {type(sirali_tuple)}")
````

````python
t1 = (1, 2, 3)
t2 = (4, 5, 6)

# Tuple birleştirme
t3 = t1 + t2
print(t3)

# Çoğaltma
t4 = t1 * 3
print(t4)

# Eleman ekleme
# Aslında yeni bir tuple oluşturuyoruz
meyveler = ("elma", "armut")
meyveler_muz = meyveler + ("muz", ) # Virgülü unutma
print(meyveler_muz)
print(meyveler)
````

## Bölüm 5 — Tuple Ne Zaman Kullanılır?

### Kullanım alanları:

**1. Sabit veriler** — değişmemesi gereken bilgiler  
**2. Fonksiyondan birden fazla değer döndürme**  
**3. Dictionary anahtarı olarak** (liste anahtar olamaz!)  
**4. Performans** — tuple, listeden daha hızlı

````python
# KULLANIM 1: SABİT VERİLER
HAFTANIN_GUNLERI = ("Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma", "Cumartesi", "Pazar")
AYLAR = ("Ocak", "Şubat", "Mart", "Nisan", "Mayıs", "Haziran", "Temmuz", "Ağustos", "Eylül", "Ekim", "Aralık")
print(f"Bugün: {HAFTANIN_GUNLERI[4]}")
print(f"Bu ay: {AYLAR[6]}")
print()

# KULLANIM 2: FONKSİYONDAN BİRDEN FAZLA DEĞER DÖNDÜRME
def istatistik_hesapla(sayilar):
    """
        Bir sayı listesinin istatistiklerini döndürür.
    """
    toplam = sum(sayilar)
    ortalama = toplam / len(sayilar)
    en_buyuk = max(sayilar)
    en_kucuk = min(sayilar)
    return toplam, ortalama, en_buyuk, en_kucuk # Tuple döndürecek

notlar = [85, 92, 95, 67, 53]
sonuc = istatistik_hesapla(notlar)
print(f"Sonuç: {sonuc}")
print()
# Unpacking ile alalım
top, ort, enb, enk = istatistik_hesapla(notlar)
print(f"Toplam: {top}")
print(f"Ortalama: {ort}")
print(f"En Büyük: {enb}")
print(f"En Küçük: {enk}")
````

````python
# KULLANIM 3: DICTIONARY ANAHTARI OLARAK
# LİSTE ANAHTAR OLAMAZ, TUPLE OLABİLİR!

# Koordinat bazlı harita
sehirler = {
    (41.01, 28.98): 'İstanbul',
    (39.57, 32.53): 'Ankara',
    (38.42, 27.14): 'İzmir'
}
print(f"(41.01, 28.98) -> {sehirler[(41.01, 28.98)]}")

# Neden liste anahtar olamıyor?
try: 
    hatali = {[1, 2]: 'test'}
except TypeError as e:
    print(f"\nListe anahtar olamaz: {e}")
print()

# KULLANIM 4: PERFORMANS KARŞILAŞTIRMASI
import sys
liste = [1, 2, 3, 4, 5]
tup = (1, 2, 3, 4, 5)
print(f"Liste Boyutu: {sys.getsizeof(liste)} byte")
print(f"Tuple Boyutu: {sys.getsizeof(tup)} byte")
````

## Bölüm 6 — Set (Küme) Nedir?

**Set** = Sırasız, tekrarsız elemanlardan oluşan veri yapısı.

| Özellik | Liste | Tuple | Set |
|---------|-------|-------|-----|
| Yazım | `[1,2,3]` | `(1,2,3)` | `{1,2,3}` |
| Sıralı mı? | Evet | Evet | Hayır |
| Tekrar var mı? | Evet | Evet | Hayır |
| Değiştirilebilir mi? | Evet | Hayır | Evet* |
| Index erişimi? | `liste[0]` | `tuple[0]` | Yok! |

\* Set'e eleman ekleyebilir/silebilirsin ama index ile erişemezsin.

Set'in matematikteki karşılığı **küme**dir:
- Bir eleman ya vardır ya yoktur — tekrar edemez
- Elemanların sırası önemli değil

## Bölüm 9 — Set Operasyonları: Birleşim, Kesişim, Fark

Set'in en güçlü tarafı **matematiksel küme operasyonları:**

| Operasyon | Metot | Operatör | Açıklama |

| Birleşim | `union()` | `\|` | Her iki kümedeki tüm elemanlar |

| Fark | `difference()` | `-` | Birinde olup diğerinde olmayan |
| Simetrik Fark | `symmetric_difference()` | `^` | Sadece birinde olan (ortak olmayanlar) |

## Bölüm 10 — frozenset: Değişmez Küme

**frozenset** = Set'in değiştirilemez (immutable) versiyonu.

| Özellik | set | frozenset |
|---------|-----|----------|
| Değiştirilebilir | Evet | Hayır |
| add/remove | Var | Yok |
| Dict anahtarı | Olamaz | Olabilir |
| Küme operasyonları | Var | Var |

Tuple, listenin immutable hali → frozenset de set'in immutable hali.
