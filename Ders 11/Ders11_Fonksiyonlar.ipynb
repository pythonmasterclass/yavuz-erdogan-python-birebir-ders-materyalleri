## Bölüm 1 — Fonksiyon Nedir? Neden Lazım?

**Problem:** Aynı kodu tekrar tekrar yazıyorsun.

```python
# Aynı selamlama 3 kez...
print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)

# ... birkaç satır sonra yine...
print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)
```

**Çözüm:** Bu kodu bir fonksiyona koy, isim ver, istediğin yerde çağır.

**Fonksiyonların 3 büyük faydası:**
1. **Tekrar kullanım** — Bir kez yaz, istediğin kadar çağır
2. **Düzen** — Kodun okunabilir ve organize olur
3. **Bakım** — Değişiklik yapman gerekirse tek yerde yap, her yere yansısın

````python
# Fonksiyonlar olmadan, tekrar eden kod
print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)

print() # Boşluk

print("=" * 30)
print("Hoş geldiniz!")
print("=" * 30)
````

````python
# Fonksiyonlar kullanılarak
def selamla():
    print("=" * 30)
    print("Hoş geldiniz!")
    print("=" * 30)
selamla() # Fonksiyonumuzun çağrısı
print() # Boş satır
selamla() # Fonksiyonumuzun çağrısı
print() # Boş satır
selamla() # Fonksiyonumuzun çağrısı
print() # Boş satır
selamla() # Fonksiyonumuzun çağrısı
````

````python
def yazdir(text):
    import sys
    sys.stdout.write(f"{text}\n")

yazdir('Merhaba Dünya')
````

## Bölüm 2 — `def` ile Fonksiyon Tanımlama

```python
def fonksiyon_adi():
    # Fonksiyonun gövdesi
    # Bu satırlar çağrıldığında çalışır
```

**Kurallar:**
- `def` anahtar kelimesi ile başlar
- Fonksiyon adı küçük harfle, `snake_case` ile yazılır
- Adın sonuna `()` ve `:` konur
- Gövde girintili (indented) yazılır
- **Tanımlamak ≠ Çalıştırmak!** `def` sadece tanımlar, çağırmak lazım.

````python
# Tanımlama
def merhaba_de():
    print("Merhaba Dünya!")
    print("Python Öğreniyorum :)")
merhaba_de()
````

````python
def cizgi_ciz():
    print('-' * 40)
cizgi_ciz()
````

## Bölüm 3 — Parametreli Fonksiyonlar: Fonksiyona Veri Gönder

Şimdiye kadarki fonksiyonlar hep aynı şeyi yapıyordu.  
Peki ya her çağrıda farklı bir isimle selamlama yapmak istersek?

**Parametre** = Fonksiyona dışarıdan gönderilen değer.

```python
def fonksiyon_adi(parametre1, parametre2):
    # parametre1 ve parametre2 burada kullanılabilir
```

````python
def selamla1():
    print('Merhaba')
selamla1()
````

````python
# Tek parametreli
def selamla2(isim):
    print(f"Merhaba {isim}. Nasılsın?")

selamla2('Yavuz')
selamla2('Efekan')
````

````python
# Birden fazla parametreli
def selamla3(isim, yas):
    print(f"Merhaba {isim}. Nasılsın?")
    if yas >= 18:
        print('Reşitsiniz.')
    else:
        print('Henüz çocuksunuz.')
selamla3('Yavuz', 20)
````

````python
# Parametreler ile hesaplama
def alan_hesapla(en, boy):
    alan = en * boy
    print(f"Dikdörtgenin alanı: {en} x {boy} = {alan} m²")

alan_hesapla(2, 10)
alan_hesapla(6, 7)
alan_hesapla(123, 987)
````

## Bölüm 4 — `return`: Fonksiyondan Değer Döndür

Şimdiye kadar fonksiyonlar sadece `print()` ile ekrana yazdırdı.  
Peki ya hesapladığı sonucu **geri göndermesini** istersek?

`return` = Fonksiyonun hesapladığı değeri dışarıya verir.

**Analoji:**
- `print` → Garson yemeği pişirip sana gösterir ama vermez
- `return` → Garson yemeği pişirip tabağına koyar, sen istersin yap

```python
def fonksiyon(parametre):
    sonuc = ... hesaplama ...
    return sonuc    # Sonucu geri gönder
```

````python
def topla1(a, b):
    print(a + b)
sonuc1 = topla1(3, 5) # 8 yazdıracak
print(sonuc1) 
````

````python
def topla2(a, b):
    return a + b
sonuc2 = topla2(3, 5) # 8 yazdıracak
print(sonuc2)

print(sonuc2 * 5)
````

````python
def daire_alani(yaricap):
    pi = 3.14159
    return pi * (yaricap ** 2)
sayi = float(input('Lütfen bir yarıçap giriniz:'))
alan = daire_alani(sayi)
print(f"Yarıçapı {sayi} olan dairenin alanı: {alan:.2f}")
````

````python
print('-' * 35)
print('HARF NOTU HESAPLAMA ARACI')
print('-' * 35)
def harf_notu(puan):
    if puan >= 90:
        return 'AA'
    elif puan >= 80:
        return 'BA'
    elif puan >= 70:
        return 'BB'
    elif puan >= 60:
        return 'CB'
    elif puan >= 50:
        return 'CC'
    else:
        return 'FF'
while True:
    print('İşlemler')
    print('1. Not Girişi')
    print('2. Çıkış')
    secim = input("Lütfen Yapmak İstediğiniz İşlemin Numarasını Giriniz")
    if secim == '2':
        print('Çıkış yapılıyor...')
        break
    else:
        ogrenci_puani = float(input('Lütfen Sınavdan Aldığınız Notu Giriniz:'))
        harf = harf_notu(ogrenci_puani)
        print(f"{ogrenci_puani} -> {harf}")
````

## Bölüm 5 — Varsayılan Parametreler: Opsiyonel Değerler

Bazen bir parametrenin çoğu zaman aynı değeri almasını istersin.  
**Varsayılan parametre** ile bunu sağlarsın:

```python
def fonksiyon(zorunlu, opsiyonel="varsayılan_değer"):
    ...
```

- Çağırırken `opsiyonel` verilirse → verilen değer kullanılır
- Verilmezse → varsayılan değer kullanılır

## Bölüm 8 — Kapsam (Scope): Değişken Nerede Geçerli?

Fonksiyon içinde tanımlanan değişkenler **sadece o fonksiyonun içinde** geçerlidir.  
Buna **yerel değişken** (local variable) denir.

| Tür | Nerede Tanımlanır | Nerede Geçerli |
|-----|------------------|----------------|
| **Yerel (local)** | Fonksiyon içinde | Sadece o fonksiyon içinde |
| **Global** | Fonksiyon dışında | Her yerde |
