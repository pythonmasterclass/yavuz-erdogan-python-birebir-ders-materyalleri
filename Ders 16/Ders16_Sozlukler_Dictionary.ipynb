## Bölüm 1 — Sözlük Nedir? Liste ile Farkı Ne?

**Dictionary** = Anahtar-değer (key-value) çiftlerinden oluşan veri yapısı.

| Özellik | Liste | Sözlük |
|---------|-------|--------|
| Erişim | Index ile (`liste[0]`) | Anahtar ile (`sozluk["isim"]`) |
| Sıra | Sıralı (0, 1, 2…) | Anahtar isimleriyle |
| Yazım | `[değer1, değer2]` | `{anahtar: değer, anahtar: değer}` |
| Ne zaman? | Sıralı veri (notlar, isimler) | Etiketli veri (isim→not, ürün→fiyat) |

```python
# Liste — index ile eriş
ogrenci_liste = ["Yavuz", 20, "İstanbul"]
# ogrenci_liste[0] → "Yavuz" ama 0 ne demek? Hatırlamak zor!

# Sözlük — isimle eriş
ogrenci_sozluk = {"isim": "Yavuz", "yas": 20, "sehir": "İstanbul"}
# ogrenci_sozluk["isim"] → "Yavuz" çok daha anlaşılır!
```

````python
# Liste ile - index ne demek hatırlamak lazım.
ogrenci_liste = ["Yavuz", 20, "Konya"]
print(f"İsim: {ogrenci_liste[0]}")
print(f"Yaş: {ogrenci_liste[1]}")
print(f"Şehir: {ogrenci_liste[2]}")
print('-' * 30)
# Sözlük ile - anahtar adı her şeyi açıklar
ogrenci_sozluk = {"isim": "Yavuz", "yas": 20, "sehir": "Konya"}
print(f"İsim: {ogrenci_sozluk["isim"]}")
print(f"Yaş: {ogrenci_sozluk["yas"]}")
print(f"Şehir: {ogrenci_sozluk["sehir"]}")
````

## Bölüm 3 — Ekleme, Güncelleme ve Silme

Sözlükler **değiştirilebilir** (mutable) — eleman ekleyebilir, güncelleyebilir, silebilirsin.

````python
# Ekleme
kisi = {"isim": "Yavuz", "yas": 20}
print(f"'sehir' ve 'bolum' eklemeden önce 'kisi': {kisi}")
kisi['sehir'] = 'Konya' # Yeni anahtar -> eklenir
kisi['bolum'] = 'Bilgisayar Mühendisliği' # Yeni anahtar -> eklenir
print(f"'sehir' ve 'bolum' eklendikten sonra 'kisi': {kisi}")
````

````python
# Güncelleme
kisi = {"isim": "Yavuz", "yas": 20, "sehir": "Konya"}
print(f"'yas' ve 'sehir' değişmeden önce 'kisi': {kisi}")
kisi['yas'] = 21
kisi['sehir'] = 'İstanbul'
print(f"'yas' ve 'sehir' değiştikten sonra 'kisi': {kisi}")
````

````python
# Silme
kisi = {"isim": "Yavuz", "yas": 20, "sehir": "Konya", "bolum": "Bilgisayar Mühendisliği"}
print(f"'bolum' silinmeden önce 'kisi': {kisi}")

# 1. Yöntem: del - anahtarı siler
del kisi["bolum"]
print(f"del sonrası 'kisi': {kisi}")

# 2. Yöntem: pop() - sil ve silinen değeri döndür
silinen_sehir = kisi.pop("sehir")
print(f".pop('sehir') sonrası 'kisi': {kisi}")
print(f"Silinen şehir: {silinen_sehir}")

# 3. Yöntem: clear() - tüm sözlüğü temizler
kisi.clear()
print(kisi)
````

````python
# pop() güvenli kullanım - anahtar yoksa varsayılan değer döndürür
kisi = {"isim": "Yavuz", "yas": 20}
# Olmayan anahtarı pop ile silmeye çalışalım
sonuc = kisi.pop("telefon", "Anahtar bulunamadı")
print(f"Sonuç: {sonuc}")
````

## Bölüm 4 — Sözlük Metotları

| Metot | Ne Yapar | Döndürdüğü |
|-------|----------|------------|
| `.keys()` | Tüm anahtarları verir | Anahtar listesi |
| `.values()` | Tüm değerleri verir | Değer listesi |
| `.items()` | Tüm çiftleri verir | (anahtar, değer) çiftleri |
| `.get(key)` | Güvenli erişim | Değer veya None |
| `.pop(key)` | Sil ve döndür | Silinen değer |
| `.update(dict2)` | Başka sözlükle güncelle | None |
| `key in sozluk` | Anahtar var mı? | True / False |

````python
notlar = {"Yavuz": 85, "Zeynep": 95, "Ege": 78, "Fatma": 68}

# .keys() → Tüm anahtarları döndürür.
print(f"Öğrenciler: {list(notlar.keys())}")

# .values() → Tüm değerleri döndürür.
print(f"Notlar: {list(notlar.values())}")

# .items() → Tüm anahtar-değer çiftlerini döndürür.
print(f"Anahtar-Değer Çiftleri: {list(notlar.items())}")
````

````python
# in operatörü → Anahtar var mı?
notlar = {"Yavuz": 85, "Zeynep": 95, "Ege": 78}

print("Yavuz" in notlar) # True
print(68 in notlar.values()) # False
print(85 in notlar.values()) # True
print(notlar["Yavuz"] == 85) # True
print("Mehmet" in notlar) # False

# Kontrol edip erişme
aranan = "Ege"
if aranan in notlar:
    print(f"{aranan}'nin notu: {notlar[aranan]}")
else:
    print(f"{aranan} bulunamadı.")
````

````python
# .update() → Başka sözlüğe güncelleme.
kisi = {"isim": "Yavuz", "yas": 22}
ek_bilgi = {"sehir": "Konya", "yas": 23} # Burada yaş var hem de farklı bir değerle.

print(f"Önce: {kisi}")
kisi.update(ek_bilgi) # Yeni anahtar eklenir, mevcut değerler farklıysa güncellenir.
print(f"Sonra: {kisi}")
````

## Bölüm 6 — Sözlük + Fonksiyon: Güçlü Kombinasyon

Fonksiyonlara sözlük gönderebilir, sözlük döndürebilirsin.  
Bu sayede veriler çok daha düzenli olur.

````python
# Fonksiyona sözlük gönderme
def ogrenci_bilgisi_yazdir(ogrenci):
    print(f"İsim: {ogrenci['isim']}")
    print(f"Yaş: {ogrenci['yas']}")
    print(f"Not: {ogrenci.get('not', 'Girilmemiş')}")

ogr1 = {"isim": "Yavuz", "yas": 23, "not": 85}
ogr2 = {"isim": "Mehmet", "yas": 19}
ogrenci_bilgisi_yazdir(ogr1)
print()
ogrenci_bilgisi_yazdir(ogr2)
````

````python
# Fonksiyondan sözlük döndürme
def ogrenci_olustur(isim, yas, puan):
    if puan >= 90: harf = "AA"
    elif puan >= 80: harf = "BA"
    elif puan >= 70: harf = "BB"
    elif puan >= 60: harf = "CB"
    elif puan >= 50: harf = "CC"
    else: harf = "FF"

    return {
        "isim": isim,
        "yas": yas,
        "puan": puan,
        "harf": harf,
        "durum": "Geçti" if puan >= 50 else "Kaldı" 
    }
ogr3 = ogrenci_olustur("Efekan", 26, 69)
print(ogr3)
print(f"{ogr3['isim']} → {ogr3['harf']} ({ogr3['durum']})")
````

````python
ogrenciler = []
ogrenciler.append(ogrenci_olustur("Yavuz", 23, 85))
ogrenciler.append(ogrenci_olustur("Zeynep", 24, 65))
ogrenciler.append(ogrenci_olustur("Melisa", 23, 85))
ogrenciler.append(ogrenci_olustur("Aleyna", 23, 85))
ogrenciler.append(ogrenci_olustur("Melike", 23, 85))

print()
print()
print()
print()
print(ogrenciler)
print()
print()
print()


print(f"{'İsim':<10} {'Yaş':>3} {'Puan':>5} {'Harf':>5} {'Durum':>6}")
print('-' * 35)
for ogr in ogrenciler:
    print(f"{ogr['isim']:<10} {ogr['yas']:>3} {ogr['puan']:>5} {ogr['harf']:>5} {ogr['durum']:>6}")
````

## Bölüm 7 — Sözlük ile Sayma ve Gruplama

Sözlüklerin en yaygın kullanımlarından biri: bir şeylerin kaç kez geçtiğini saymak.  
Önceki ödevde bunu paralel listelerle yapmıştık — sözlükle çok daha kolay!

````python
# Harf sayma — bir metindeki her harfin kaç kez geçtiğini bul
metin = "missisipippi"
sayac = {}

for harf in metin:
    if harf in sayac:
        sayac[harf] += 1 # Varsa 1 arttır
    else:
        sayac[harf] = 1 # Yoksa 1 ile başlat

print(f"Metin: {metin}")
for harf, adet in sayac.items():
    print(f"'{harf}' -> {adet} kez")
````

````python
# Kelime sayma
cumle = "python çok güzel bir dil python ile her şey mümkün python harika"
kelimeler = cumle.split()
kelime_sayac = {}
for kelime in kelimeler:
    if kelime in kelime_sayac:
        kelime_sayac[kelime] += 1
    else:
        kelime_sayac[kelime] = 1
print("Kelime sıklıkları")
for kelime, adet in kelime_sayac.items():
    print(f"'{kelime}' -> {adet} kez")
````

````python
# Oy sayma — TR Simülasyonu
oylar = ["Aday A", "Aday B", "Aday A", "Aday C", "Aday B", "Aday A", "Aday A", "Aday C", "Aday B", "Aday A"]
sonuclar = {}
for oy in oylar:
    if oy in sonuclar:
        sonuclar[oy] += 1
    else:
        sonuclar[oy] = 1

print("--- Seçim Sonuçları ---")
toplam_oy = len(oylar)
for aday, oy_sayisi in sonuclar.items():
    oran = oy_sayisi / toplam_oy * 100
    print(f"{aday}: {oy_sayisi} oy (%{oran:.1f})")

en_cok_oy = 0
kazanan = ""
for aday, oy_sayisi in sonuclar.items():
    if oy_sayisi > en_cok_oy:
        en_cok_oy = oy_sayisi
        kazanan = aday
print(f"\nKazanan: {kazanan} ({en_cok_oy})")
````
