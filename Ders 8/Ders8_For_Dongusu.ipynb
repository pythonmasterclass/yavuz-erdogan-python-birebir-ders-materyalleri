## Bölüm 2 — `for` Döngüsünün Temel Yapısı

```python
for degisken in tekrarlanabilir_sey:
    # her tekrarda çalışacak kod
```

- `degisken` → Her turda farklı bir değer alan geçici değişken
- `tekrarlanabilir_sey` → Bir liste, bir range, bir string...
- `:` → Satır sonunda zorunlu
- **Girinti** → 4 boşluk (if'teki ile aynı mantık!)

````python
meyveler = ['elma', 'armut', 'muz', 'çilek']
for meyve in meyveler:
    print(meyve)
````

````python
renkler = ['kırmızı', 'mavi', 'yeşil']
for renk in renkler:
    print(f"Şu an renk: {renk}")
print("Döngü bitti")
````

## Bölüm 3 — Listelerle `for` Döngüsü

`for` döngüsünün en yaygın kullanımı: bir listedeki her elemanla bir işlem yapmak.

````python
# Elimizde içerisinde sayılar olan bir liste olsun.
# Bu listedeki her elemanın kübünü hesaplayıp ekrana yazdıralım.

sayilar = [2, 5, 8, 10, 3]
for sayi in sayilar:
    kup = sayi ** 3
    print(f"{sayi}³ = {kup}")
````

````python
# Elimizde içerisinde bazı öğrencilerin not ortalamasının olduğu bir liste olsun.
# Bu listedeki her elemana göre öğrencinin kalma-geçme durumunu kontrol ederek ekrana yazdıralım.
ortalamalar = [85, 42, 78, 95, 53, 67]
for ortalama in ortalamalar:
    if ortalama >= 50:
        print(f"{ortalama} -> Geçti")
    else:
        print(f"{ortalama} -> Kaldı")
````

## Bölüm 4 — `range()` Fonksiyonu: Sayı Dizisi Üretmek

`range()` bir sayı dizisi üretir. `for` ile birlikte "şu kadar kez tekrarla" demek için kullanılır.

**3 farklı kullanımı:**

| Kullanım | Anlamı | Ürettiği Sayılar |
|----------|--------|------------------|
| `range(5)` | 0'dan 5'e kadar (5 hariç) | 0, 1, 2, 3, 4 |
| `range(2, 7)` | 2'den 7'ye kadar (7 hariç) | 2, 3, 4, 5, 6 |
| `range(0, 10, 2)` | 0'dan 10'a kadar 2'şer | 0, 2, 4, 6, 8 |

````python
# range(5) -> 5'e kadar say.
for i in range(5):
    print(i)
````

````python
# range(1, 6) -> 1'den başla 6'ya kadar say.
for i in range(1, 6):
    print(i)
````

````python
# range(0, 11, 2) -> 0'dan başla 11'e kadar 2'şer 2'şer say.
for i in range(0, 11, 2):
    print(i)
````

````python
# range(10, 0, -1) -> 10'dan başla 0'a kadar -1 eksilerek say.
for i in range(10, 0, -1):
    print(i)
````

````python
# ALIŞTIRMA: 5'in katlarını yazdır (5, 10, 15, 20)
for i in range(5, 21, 5):
    print(i)
````

## Bölüm 5 — Döngüyle Liste Oluşturma, Toplama ve Sayma

Döngünün en güçlü kullanımlarından biri: bir listeyi **gezerken** yeni bilgi üretmek.

Temel kalıplar:
- **Toplama:** Bir sayaçla topla
- **Sayma:** Bir koşula uyanları say
- **Yeni liste:** Boş listeye `append` ile ekle

````python
# Kalıp 1: Döngüyle toplama yapma
sayilar = [5, 55, 83, 74, 98, 110]

toplam = 0
# sum() kullanmadan döngüyle toplama yapma
for sayi in sayilar:
    toplam += sayi

print(toplam)
````

````python
# Kalıp 2: Döngüyle sayma yapma
notlar = [35,65,80,28,38,34,26,54,59,86,79]
gecen_sayisi = 0
kalan_sayisi = 0
for not_puan in notlar:
    if not_puan >= 50:
        gecen_sayisi += 1
    else:
        kalan_sayisi += 1
print(gecen_sayisi)
print(kalan_sayisi)
````

````python
# Kalıp 3: Döngüyle yeni liste oluşturma
notlar = [35,65,80,28,38,34,26,54,59,86,79]
gecenler = []
kalanlar = []
for not_puan in notlar:
    if not_puan >= 50:
        gecenler.append(not_puan)
    else:
        kalanlar.append(not_puan)


print('=' * 45)
print('12-A MATEMATİK DERSİ')
print('SINAV İSTATİSTİKLERİ')
print('-' * 45)
print(f"Geçen Kişi Sayısı: {len(gecenler)}")
print(f"Kalan Kişi Sayısı: {len(kalanlar)}")
print('-' * 45)
print(f"Geçen Notlar: {gecenler}")
print(f"Kalan Notlar: {kalanlar}")
print('=' * 45)
````

## Bölüm 6 — `range()` + `len()` ile İndeksli Döngü

Bazen listeyi gezerken sadece elemana değil, **indekse** de ihtiyaç duyarsın.  
Özellikle paralel listelerle çalışırken.

````python
# Paralel listeler - indeksle erişme
ogrenciler = ['Ali', 'Ayşe', 'Mehmet', 'Zeynep', 'Can']
notlar = [20, 15, 67, 98, 100]

for i in range(len(ogrenciler)):
    print(f"{ogrenciler[i]:10s} -> {notlar[i]} puan")    
````

````python
# Sıra numarası ile yazdırma
sehirler = ['İstanbul', 'Ankara', 'İzmir']
for a in range(len(sehirler)):
    print(f"{a + 1}. {sehirler[a]}")
````

## Bölüm 7 — `enumerate()`: İndeks + Değer Birlikte

`range(len(...))` yerine daha temiz bir yol var: `enumerate()`

Her turda hem **indeksi** hem **değeri** birlikte verir.

````python
meyveler = ['elma', 'armut', 'muz', 'çilek']
for indeks, meyve in enumerate(meyveler):
    print(indeks, meyve)
````

````python
meyveler = ['elma', 'armut', 'muz', 'çilek']
for indeks, meyve in enumerate(meyveler, start = 1):
    print(indeks, meyve)
````

````python
# Paralel listelerde enumerate kullanımı
ogrenciler = ['Ali', 'Ayşe', 'Mehmet', 'Zeynep', 'Can']
notlar = [20, 15, 67, 98, 100]

for i, o in enumerate(ogrenciler):
    print(f"{o} -> {notlar[i]} puan")    
````

## Bölüm 10 — Uygulama 1: Sınıf Karne Sistemi

Geçen derslerdeki paralel listeler + bugünkü for döngüsü = güçlü uygulama!

````python
ogrenciler = ['Ali', 'Ayşe', 'Mehmet', 'Zeynep', 'Can', 'Elif', 'Efekan']
notlar = [78, 92, 45, 88, 67, 95, 53]

# Başlık
print('=' * 45)
print('            SINIF KARNE RAPORU')
print('=' * 45)
print(f"{'Sıra':<6}{'Öğrenci':<12}{'Not':<8}{'Durum'}")
print('-' * 45)
gecen = 0
kalan = 0

for i, ogrenci in enumerate(ogrenciler, start=1):
    not_puani = notlar[i-1]
    # Harf notu
    if not_puani >= 90:
        harf = 'AA'
    elif not_puani >= 80:
        harf = 'BA'
    elif not_puani >= 70:
        harf = 'BB'
    elif not_puani >= 60:
        harf = 'CC'
    else:
        harf = 'FF'

    # Durum
    if not_puani >= 50:
        durum = 'Geçti'
        gecen += 1
    else:
        durum = 'Kaldı'
        kalan += 1

    print(f"{i:<6}{ogrenci:<12}{not_puani:<8}{durum}")

# İstatistikler
print('=' * 45)
print(f"Sınıf Ortalaması: {sum(notlar) / len(notlar) :.2f}")
print(f"En Yüksek Not: {max(notlar)}")
print(f"En Düşük Not: {min(notlar)}")
print(f"Geçen Kişi Sayısı: {gecen}")
print(f"Kalan Kişi Sayısı: {kalan}")
print(f"Geçme Oranı: %{(gecen/len(ogrenciler))*100:.0f}")
print('=' * 45)
````

## Bölüm 11 — Uygulama 2: Geliştirilmiş Harcama Takipçisi

````python
kategoriler = ['Barınma', 'Gıda', 'Ulaşım', 'Faturalar', 'Eğlence', 'Diğer']
harcamalar = []
print('=' * 30)
print('Lütfen aylık harcamalarınızı giriniz ↓')

# 4.derste 6 kez yazdığımız kodu tek döngüye indirgeyelim
for kategori in kategoriler:
    harcama = float(input(f"{kategori}: "))
    harcamalar.append(harcama)

gelir = float(input("Aylık Geliriniz (TL):"))

# Rapor
print()
print("=" * 30)
print(f'{"AYLIK FİNANS RAPORU":^30}')
print("=" * 30)

for sira, kategori in enumerate(kategoriler):
    print(f"{kategori:15s} {harcamalar[sira]:>10.2f} TL")

print("-" * 35)
toplam = sum(harcamalar)
print(f"{'TOPLAM':15s} {toplam:>10.2f} TL")
print(f"{'ORTALAMA':15s} {toplam/len(harcamalar):>10.2f} TL")

en_yuksek_harcama_indeksi = harcamalar.index(max(harcamalar))
en_dusuk_harcama_indeksi = harcamalar.index(min(harcamalar))

print(f"En Fazla Harcama Yapılan Kategori: {kategoriler[en_yuksek_harcama_indeksi]} ({max(harcamalar):.2f}) TL)")
print(f"En Az Harcama Yapılan Kategori: {kategoriler[en_dusuk_harcama_indeksi]} ({min(harcamalar):.2f}) TL)")

fark = gelir - toplam
if fark >= 0:
    print(f"\nKalan: {fark:,.2f} TL -> Bütçemiz dengede! Harika ilerliyorsun.")
else:
    print(f"\nAçık: {abs(fark):,.2f} TL -> Dikkat, bütçe açıldı!")

# Hangi kategoriler ortalamanın üzerinde onu yazdıralım
ortalama = toplam / len(harcamalar)
print(f'\nOrtalamanın ({ortalama:.2f} TL) Üzerindeki Kategoriler ↓')
for sira, kategori in enumerate(kategoriler):
    if harcamalar[sira] > ortalama:
        print(f"→ {kategori}: {harcamalar[sira]:.2f} TL")
print("=" * 35)
````

## Bölüm 12 — Uygulama 3: Basit Şifre Gücü Kontrol

````python
sifre = input('Bir şifre giriniz:')

buyuk_harf = 0
kucuk_harf = 0
rakam = 0
ozel = 0

buyuk_harfler = "ABCÇDEFGĞHIİJKLMNOÖPRSŞTUÜVYZ"
kucuk_harfler = "abcçdefgğhıijklmnoöprsştuüvyz"
rakamlar = "0123456789"

for karakter in sifre:
    if karakter in buyuk_harfler:
        buyuk_harf += 1
    elif karakter in kucuk_harfler:
        kucuk_harf += 1
    elif karakter in rakamlar:
        rakam += 1
    else:
        ozel += 1

print(f"\nŞifre: {sifre}")
print(f"Uzunluk: {len(sifre)}")
print(f"Büyük Harf Sayısı: {buyuk_harf}")
print(f"Küçük Harf Sayısı: {kucuk_harf}")
print(f"Rakam Sayısı: {rakam}")
print(f"Özel Karakter Sayısı: {ozel}")

guc = 0
# Şifremiz 8 karakterden uzun mu?
# Büyük harf var mı? Varsa 1 puan arttıralım.
# Küçük harf var mı? Varsa 1 puan arttıralım.
# Rakam var mı? Varsa 1 puan arttıralım.
# Özel karakter var mı? Varsa 1 puan arttıralım.

if len(sifre) >= 8:
    guc += 1
if buyuk_harf > 0:
    guc += 1
if kucuk_harf > 0:
    guc += 1
if rakam > 0:
    guc += 1
if ozel > 0:
    guc += 1

print(f'Güç Puanı: {guc}')
if guc >= 4:
    print('Güçlü şifre :)')
elif guc >= 2:
    print('Biraz daha iyi olabilir ancak bu da iyi :|')
else:
    print('ruhi123 mü şifren amk?')
````

````python
# ========================================
# MİNİ UYGULAMA: Quiz v2 — Kendin yaz!
# İpucu: sorular ve cevaplar listede,
# for döngüsüyle her soruyu sor
# ========================================

````

### Örnek Çözüm

````python
# ========================================
# MİNİ UYGULAMA: Quiz v2 — Örnek Çözüm
# ========================================

sorular = [
    "Python'da liste hangi parantezle oluşturulur?",
    "Eleman sayısını veren fonksiyon?",
    "Python'da sayma kaçtan başlar?",
    "input() her zaman hangi tip döndürür?",
    "if bloğunda kaç boşluk girinti kullanılır?",
    "Listeye sona eleman ekleyen metot?",
    "En büyük değeri bulan fonksiyon?",
    "== operatörü ne kontrol eder?"
]

cevaplar = ["[]", "len", "0", "str", "4", "append", "max", "esitlik"]

puan = 0
yanlis_sorular = []

print("=" * 45)
print(f"      PYTHON QUIZ ({len(sorular)} Soru)")
print("=" * 45)

for i, soru in enumerate(sorular, start=1):
    print(f"\nSoru {i}: {soru}")
    cevap = input("Cevabın: ")
    
    if cevap.lower() == cevaplar[i - 1].lower():
        print("Doğru! ✓")
        puan = puan + 1
    else:
        print(f"Yanlış. Doğru cevap: {cevaplar[i - 1]}")
        yanlis_sorular.append(i)

# Sonuç raporu
print("\n" + "=" * 45)
yuzde = (puan / len(sorular)) * 100
print(f"Puanın: {puan} / {len(sorular)} (%{yuzde:.0f})")

if yuzde == 100:
    print("Mükemmel! Hepsini bildin!")
elif yuzde >= 75:
    print("Harika! Çok iyi gidiyorsun.")
elif yuzde >= 50:
    print("İyi! Biraz daha pratik yap.")
else:
    print("Konuları tekrar gözden geçir.")

if len(yanlis_sorular) > 0:
    print(f"Yanlış yaptığın sorular: {yanlis_sorular}")

print("=" * 45)
````
